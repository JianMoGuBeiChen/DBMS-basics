```python
import csv
import mysql.connector

# 1. Connect to the MySQL database
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="@DemonKingLuo1",
    database="dbms"
)
cursor = conn.cursor()

# 2. Create the table (Column names with spaces/dots wrapped in backticks)
cursor.execute('''
    CREATE TABLE IF NOT EXISTS crime_data (
        `Crime Id` INT,
        `Original Crime Type Name` VARCHAR(255),
        `Report Date` DATE,
        `Call Date` DATE,
        `Offense Date` DATE,
        `Call Time` TIME,
        `Call Date Time` DATETIME,
        `Disposition` VARCHAR(255),
        `Address` VARCHAR(255),
        `City` VARCHAR(255),
        `State` VARCHAR(255),
        `Address Type` VARCHAR(255),
        `Case Status` VARCHAR(255)
    )
''')

csv_file_path = 'police_data_filtered.csv'

# 3. Read the CSV and prepare data for bulk insertion
with open(csv_file_path, mode='r', encoding='utf-8') as file:
    csv_reader = csv.reader(file)
    
    # Skip the header row if your CSV has one
    next(csv_reader)
    
    # Extract all rows into a list of tuples
    data_to_insert = [tuple(row[1:]) for row in csv_reader]

# 4. Use executemany to insert all rows efficiently (Backticks added here too)
insert_query = '''
    INSERT INTO crime_data (
        `Crime Id`, `Original Crime Type Name`, `Report Date`, 
        `Call Date`, `Offense Date`, `Call Time`, `Call Date Time`, 
        `Disposition`, `Address`, `City`, `State`, 
        `Address Type`, `Case Status`
    ) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
'''
cursor.executemany(insert_query, data_to_insert)

# 5. Commit the transaction and close the connection
conn.commit()
print(f"Successfully inserted {cursor.rowcount} rows.")

cursor.close()
conn.close()
```
